## Divisione intera e modulo

Ricorda che l’operatore di divisione intera, `//`, divide due numeri e arrotonda il risultato per difetto a un intero. Supponiamo, per esempio, che un film duri 105 minuti e di voler sapere quante ore sono. La divisione ordinaria restituisce un numero in virgola mobile:

```python
minutes = 105
minutes / 60
```

Ma di solito non esprimiamo le ore con cifre decimali. La divisione intera restituisce il numero intero di ore, arrotondato per difetto:

```python
minutes = 105
hours = minutes // 60
hours
```

Per ottenere il resto, potresti sottrarre un’ora espressa in minuti:

```python
remainder = minutes - hours * 60
remainder
```

Oppure puoi usare l’**operatore modulo**, `%`, che divide due numeri e restituisce il resto.

```python
remainder = minutes % 60
remainder
```

L’operatore modulo è più utile di quanto possa sembrare. Per esempio, permette di verificare se un numero è divisibile per un altro: se `x % y` è zero, `x` è divisibile per `y`.

Può anche estrarre l’ultima cifra, o le ultime cifre, di un numero. Per esempio, `x % 10` restituisce l’ultima cifra di `x` (in base 10); analogamente, `x % 100` restituisce le ultime due cifre.

```python
x = 123
x % 10
```

```python
x % 100
```

Infine, l’operatore modulo permette di fare «aritmetica dell’orologio». Per esempio, se un evento inizia alle 11 del mattino e dura tre ore, possiamo usare il modulo per determinare a che ora termina.

```python
start = 11
duration = 3
end = (start + duration) % 12
end
```

L’evento terminerebbe alle 2 del pomeriggio.

## Operatori logici

Per combinare valori booleani in espressioni possiamo usare gli **operatori logici**. I più comuni sono `and`, `or` e `not`. Il loro significato è simile a quello delle corrispondenti parole inglesi. Per esempio, l’espressione seguente vale `True` soltanto se `x` è maggiore di `0` **e** minore di `10`.

```python
x > 0 and x < 10
```

L’espressione seguente vale `True` se **almeno una** delle condizioni è vera, cioè se il numero è divisibile per 2 **oppure** per 3:

```python
x % 2 == 0 or x % 3 == 0
```

Infine, `not` nega un’espressione booleana: l’espressione seguente vale `True` se `x > y` è `False`.

```python
not x > y
```

A rigore, gli operandi di un operatore logico dovrebbero essere espressioni booleane, ma Python non è così rigido. Qualsiasi numero diverso da zero viene interpretato come `True`:

```python
42 and True
```

Questa flessibilità può essere utile, ma presenta sottigliezze che possono creare confusione. Potresti preferire di evitarla.

## Istruzioni `if`

Per scrivere programmi utili dobbiamo quasi sempre poter verificare condizioni e modificare di conseguenza il comportamento del programma. Le **istruzioni condizionali** ci permettono di farlo. La forma più semplice è `if`:

```python
if x > 0:
    print('x is positive')
```

`if` è una parola chiave di Python. Le istruzioni `if` hanno la stessa struttura delle definizioni di funzione: un’intestazione seguita da una o più istruzioni rientrate, chiamate **blocco**.

L’espressione booleana dopo `if` è la **condizione**. Se è vera, le istruzioni del blocco vengono eseguite; altrimenti no.

Non c’è un limite al numero di istruzioni nel blocco, ma deve essercene almeno una. Talvolta è utile un blocco che non faccia nulla, per riservare il posto a codice ancora da scrivere. In tal caso si usa `pass`, che non esegue alcuna operazione.

```python
if x < 0:
    pass          # TODO: need to handle negative values!
```

La parola `TODO` in un commento è una convenzione per ricordare che resta qualcosa da fare.

## La clausola `else`

Un’istruzione `if` può avere una seconda parte, detta clausola `else`. La sintassi è questa:

```python
if x % 2 == 0:
    print('x is even')
else:
    print('x is odd')
```

Se la condizione è vera, viene eseguita la prima istruzione rientrata; altrimenti viene eseguita la seconda.

In questo esempio, se `x` è pari, il resto della divisione per `2` è `0`: la condizione è vera e il programma mostra `x is even`. Se `x` è dispari, il resto è `1`: la condizione è falsa e il programma mostra `x is odd`.

Poiché la condizione è necessariamente vera o falsa, viene eseguita esattamente una delle alternative. Le alternative si chiamano **rami**.

## Condizioni concatenate

Talvolta le possibilità sono più di due e servono più rami. Un modo per esprimere un calcolo di questo tipo è una **condizione concatenata**, che include una clausola `elif`.

```python
if x < y:
    print('x is less than y')
elif x > y:
    print('x is greater than y')
else:
    print('x and y are equal')
```

`elif` è l’abbreviazione di «else if». Non c’è un limite al numero di clausole `elif`. Se è presente `else`, deve trovarsi alla fine, ma non è obbligatoria.

Le condizioni vengono controllate nell’ordine. Se la prima è falsa, si controlla la successiva, e così via. Quando una è vera, viene eseguito il ramo corrispondente e l’istruzione `if` termina. Anche se più condizioni sono vere, viene eseguito soltanto il primo ramo vero.

## Condizioni annidate

Un’istruzione condizionale può trovarsi all’interno di un’altra. Avremmo potuto scrivere l’esempio precedente così:

```python
if x == y:
    print('x and y are equal')
else:
    if x < y:
        print('x is less than y')
    else:
        print('x is greater than y')
```

L’istruzione `if` esterna contiene due rami. Il primo contiene un’istruzione semplice; il secondo contiene un altro `if`, che a sua volta ha due rami. Entrambi contengono istruzioni semplici, ma potrebbero contenere altre condizioni.

Benché il rientro renda evidente la struttura, le **condizioni annidate** possono essere difficili da leggere. Ti suggerisco di evitarle quando possibile.

Gli operatori logici spesso permettono di semplificarle. Ecco un esempio di condizione annidata.

```python
if 0 < x:
    if x < 10:
        print('x is a positive single-digit number.')
```

L’istruzione `print` viene eseguita solo se entrambe le condizioni sono soddisfatte; otteniamo quindi lo stesso effetto con `and`.

```python
if 0 < x and x < 10:
    print('x is a positive single-digit number.')
```

Per questo tipo di condizione Python offre una forma più concisa:

```python
if 0 < x < 10:
    print('x is a positive single-digit number.')
```

## Ricorsione

Una funzione può chiamare sé stessa. Non è immediatamente evidente perché sia utile, ma è una delle possibilità più sorprendenti offerte dalla programmazione. Ecco un esempio.

```python
def countdown(n):
    if n <= 0:
        print('Blastoff!')
    else:
        print(n)
        countdown(n-1)
```

Se `n` è zero o negativo, `countdown` mostra la parola «Blastoff!». Altrimenti mostra `n` e poi chiama sé stessa, passando `n-1` come argomento.

Ecco cosa succede quando la chiamiamo con argomento `3`.

```python
countdown(3)
```

L’esecuzione di `countdown` comincia con `n=3`. Poiché `n` è maggiore di `0`, mostra `3` e poi richiama sé stessa…

> L’esecuzione di `countdown` comincia con `n=2`. Mostra `2` e richiama sé stessa…
>
> > L’esecuzione di `countdown` comincia con `n=1`. Mostra `1` e richiama sé stessa…
> >
> > > L’esecuzione di `countdown` comincia con `n=0`. Poiché `n` non è maggiore di `0`, mostra «Blastoff!» e termina.
> >
> > La chiamata con `n=1` termina.
>
> La chiamata con `n=2` termina.

La chiamata con `n=3` termina.

Una funzione che chiama sé stessa è **ricorsiva**. Come altro esempio, possiamo scrivere una funzione che visualizza una stringa `n` volte.

```python
def print_n_times(string, n):
    if n > 0:
        print(string)
        print_n_times(string, n-1)
```

Se `n` è positivo, `print_n_times` mostra `string` e richiama sé stessa passando `string` e `n-1`.

Se `n` è zero o negativo, la condizione è falsa e `print_n_times` non fa nulla.

Ecco come funziona.

```python
print_n_times('Spam ', 4)
```

Per esempi semplici come questo sarebbe probabilmente più facile usare un ciclo `for`. Vedremo però esempi difficili da scrivere con `for` e facili con la ricorsione: è utile iniziare a conoscerla.

## Glossario

**ricorsione:** processo mediante il quale una funzione in esecuzione chiama sé stessa.

**operatore modulo:** operatore `%` che restituisce il resto della divisione tra numeri interi.

**espressione booleana:** espressione il cui valore è `True` oppure `False`.

**operatore relazionale:** operatore che confronta i propri operandi: `==`, `!=`, `>`, `<`, `>=`, `<=`.

**operatore logico:** operatore che combina espressioni booleane, come `and`, `or`, `not`.

**istruzione condizionale:** istruzione che controlla il flusso di esecuzione in base a una condizione.

**condizione:** espressione booleana che determina quale ramo di un’istruzione condizionale viene eseguito.

**blocco:** una o più istruzioni rientrate per indicare che fanno parte di un’altra istruzione.

**ramo:** una delle sequenze alternative di istruzioni in un’istruzione condizionale.

**condizione concatenata:** istruzione condizionale con una serie di rami alternativi.

**condizione annidata:** istruzione condizionale presente in un ramo di un’altra istruzione condizionale.

**ricorsiva:** si dice di una funzione che chiama sé stessa.

**caso base:** ramo di una funzione ricorsiva che non effettua una chiamata ricorsiva.

**ricorsione infinita:** ricorsione senza caso base o che non lo raggiunge mai; alla fine provoca un errore durante l’esecuzione.

**carattere di nuova riga:** carattere che crea un’interruzione di riga in una stringa.
