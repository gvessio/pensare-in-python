## Il modulo `jupyturtle`

Per utilizzare il modulo `jupyturtle`, possiamo importarlo in questo modo.

````python
import jupyturtle
````

Ora possiamo utilizzare le funzioni definite nel modulo, come `make_turtle` e `forward`.

````python
jupyturtle.make_turtle()
jupyturtle.forward(100)
````

`make_turtle` crea una **superficie di disegno** (*canvas*), cioè uno spazio sullo schermo in cui possiamo disegnare, e una tartaruga, rappresentata da un guscio circolare e una testa triangolare.
Il cerchio indica la posizione della tartaruga e il triangolo la direzione verso cui è rivolta.

`forward` sposta la tartaruga di una determinata distanza nella direzione verso cui è rivolta, tracciando un segmento durante lo spostamento.
La distanza è espressa in unità arbitrarie: la dimensione effettiva dipende dallo schermo del computer.

Utilizzeremo spesso le funzioni definite nel modulo `jupyturtle`, quindi sarebbe comodo non dover scrivere ogni volta il nome del modulo.
Possiamo farlo importando il modulo in quest'altro modo.

````python
from jupyturtle import make_turtle, forward
````

Questa versione dell'istruzione di importazione importa `make_turtle` e `forward` dal modulo `jupyturtle`, così possiamo chiamarle direttamente.

````python
make_turtle()
forward(100)
````

`jupyturtle` mette a disposizione altre due funzioni che utilizzeremo: `left` e `right`. Le importiamo in questo modo.

````python
from jupyturtle import left, right
````

`left` fa girare la tartaruga a sinistra. Accetta un argomento, cioè l'angolo di rotazione espresso in gradi. Per esempio, possiamo farla ruotare di 90 gradi verso sinistra in questo modo.

````python
make_turtle()
forward(50)
left(90)
forward(50)
````

Questo programma sposta la tartaruga prima verso est e poi verso nord, lasciando dietro di sé due segmenti. Prima di proseguire, prova a modificare il programma precedente in modo da disegnare un quadrato.

## Disegnare un quadrato

Ecco un modo per disegnare un quadrato.

````python
make_turtle()

forward(50)
left(90)

forward(50)
left(90)

forward(50)
left(90)

forward(50)
left(90)
````

Poiché questo programma ripete quattro volte la stessa coppia di righe, possiamo ottenere lo stesso risultato in modo più conciso con un ciclo `for`.

````python
make_turtle()
for i in range(4):
    forward(50)
    left(90)
````

## Incapsulamento e generalizzazione

Prendiamo il codice per disegnare un quadrato della sezione precedente e inseriamolo in una funzione chiamata `square`.

````python
def square():
    for i in range(4):
        forward(50)
        left(90)
````

Ora possiamo chiamare la funzione in questo modo.

````python
make_turtle()
square()
````

Racchiudere una porzione di codice in una funzione si chiama **incapsulamento** (*encapsulation*).
Uno dei vantaggi dell'incapsulamento è che associa un nome al codice, fornendo una sorta di documentazione. Un altro vantaggio è che, se riutilizziamo il codice, chiamare due volte una funzione è più conciso che copiare e incollare il suo corpo!

Nella versione attuale, il lato del quadrato misura sempre `50`.
Se vogliamo disegnare quadrati di dimensioni diverse, possiamo utilizzare la lunghezza del lato come parametro.

````python
def square(length):
    for i in range(4):
        forward(length)
        left(90)
````

Ora possiamo disegnare quadrati di dimensioni diverse.

````python
make_turtle()
square(30)
square(60)
````

Aggiungere un parametro a una funzione si chiama **generalizzazione**, perché rende la funzione più generale: con la versione precedente il quadrato ha sempre la stessa dimensione, mentre con questa può averne una qualsiasi.

Aggiungendo un altro parametro possiamo rendere la funzione ancora più generale.
La funzione seguente disegna poligoni regolari con un numero di lati specificato.

````python
def polygon(n, length):
    angle = 360 / n
    for i in range(n):
        forward(length)
        left(angle)
````

In un poligono regolare con `n` lati, l'angolo di rotazione tra due lati consecutivi è `360 / n` gradi.

L'esempio seguente disegna un poligono di `7` lati, ciascuno lungo `30` unità.

````python
make_turtle()
polygon(7, 30)
````

Quando una funzione ha più di un paio di argomenti numerici, è facile dimenticare che cosa rappresentino o in quale ordine vadano passati. Può essere utile includere i nomi dei parametri nell'elenco degli argomenti.

````python
make_turtle()
polygon(n=7, length=30)
````

Questi vengono talvolta chiamati «argomenti con nome», perché includono i nomi dei parametri.
In Python, però, sono chiamati più spesso **argomenti con parola chiave** (*keyword arguments*), da non confondere con le parole chiave di Python come `for` e `def`.

L'uso del segno `=` ci ricorda come funzionano argomenti e parametri: quando chiamiamo una funzione, i valori degli argomenti vengono assegnati ai parametri.

## Approssimare un cerchio

Supponiamo ora di voler disegnare un cerchio.
Possiamo farlo, in modo approssimato, disegnando un poligono con molti lati, ciascuno abbastanza corto da essere difficilmente distinguibile.
Ecco una funzione che usa `polygon` per disegnare un poligono di `30` lati che approssima un cerchio.

````python
import math

def circle(radius):
    circumference = 2 * math.pi * radius
    n = 30
    length = circumference / n
    polygon(n, length)
````

`circle` riceve come parametro il raggio del cerchio.
Calcola `circumference`, cioè la circonferenza di un cerchio con quel raggio.
`n` è il numero di lati, quindi `circumference / n` è la lunghezza di ciascun lato.

Questa funzione potrebbe richiedere molto tempo per essere eseguita.
Possiamo velocizzarla chiamando `make_turtle` con un argomento con parola chiave chiamato `delay`, che imposta il tempo, in secondi, durante il quale la tartaruga attende dopo ogni passo.
Il valore predefinito è `0.2` secondi: impostandolo a `0.02`, l'esecuzione diventa circa dieci volte più veloce.

````python
make_turtle(delay=0.02)
circle(30)
````

Un limite di questa soluzione è che `n` è una costante: per cerchi molto grandi i lati risultano troppo lunghi, mentre per cerchi piccoli perdiamo tempo a disegnare lati molto corti.
Una possibilità sarebbe generalizzare la funzione rendendo `n` un parametro. Per ora, però, manteniamo le cose semplici.

## Refactoring

Scriviamo ora una versione più generale di `circle`, chiamata `arc`, che accetta un secondo parametro, `angle`, e disegna un arco di cerchio corrispondente all'angolo indicato.
Per esempio, se `angle` vale `360` gradi, disegna un cerchio completo; se vale `180` gradi, disegna un semicerchio.

Per scrivere `circle` abbiamo potuto riutilizzare `polygon`, perché un poligono con molti lati è una buona approssimazione di un cerchio.
Non possiamo però utilizzare direttamente `polygon` per scrivere `arc`.

Creeremo invece una versione più generale di `polygon`, chiamata `polyline`.

````python
def polyline(n, length, angle):
    for i in range(n):
        forward(length)
        left(angle)
````

`polyline` accetta come parametri il numero di segmenti da disegnare, `n`, la loro lunghezza, `length`, e l'angolo tra di essi, `angle`.

Ora possiamo riscrivere `polygon` utilizzando `polyline`.

````python
def polygon(n, length):
    angle = 360.0 / n
    polyline(n, length, angle)
````

E possiamo utilizzare `polyline` per scrivere `arc`.

````python
def arc(radius, angle):
    arc_length = 2 * math.pi * radius * angle / 360
    n = 30
    length = arc_length / n
    step_angle = angle / n
    polyline(n, length, step_angle)
````

`arc` è simile a `circle`, ma calcola `arc_length`, cioè la lunghezza dell'arco, che è una frazione della circonferenza completa.

Infine, possiamo riscrivere `circle` utilizzando `arc`.

````python
def circle(radius):
    arc(radius,  360)
````

Per verificare che queste funzioni si comportino come previsto, le utilizzeremo per disegnare qualcosa di simile a una lumaca. Con `delay=0`, la tartaruga si muove alla massima velocità possibile.

````python
make_turtle(delay=0)
polygon(n=20, length=9)
arc(radius=70, angle=70)
circle(radius=10)
````

In questo esempio siamo partiti da codice funzionante e lo abbiamo riorganizzato usando funzioni diverse.
Modifiche di questo tipo, che migliorano il codice senza cambiarne il comportamento, sono chiamate **refactoring**.

Se avessimo pianificato tutto in anticipo, avremmo potuto scrivere subito `polyline` ed evitare il refactoring, ma spesso all'inizio di un progetto non sappiamo abbastanza per progettare tutte le funzioni.
Quando iniziamo a programmare, comprendiamo meglio il problema.
A volte il refactoring è il segno che abbiamo imparato qualcosa.

## Un piano di sviluppo

Un **piano di sviluppo** è un procedimento per scrivere programmi.
Quello utilizzato in questo capitolo è «incapsulamento e generalizzazione» e prevede questi passaggi:

1. Inizia scrivendo un piccolo programma senza definire funzioni.
2. Quando il programma funziona, individua una parte coerente del codice, racchiudila in una funzione e assegnale un nome.
3. Generalizza la funzione aggiungendo i parametri opportuni.
4. Ripeti i passaggi da 1 a 3 fino a ottenere un insieme di funzioni funzionanti.
5. Cerca opportunità per migliorare il programma attraverso il refactoring. Per esempio, se hai codice simile in più punti, valuta di raccoglierlo in una funzione sufficientemente generale.

Questo procedimento ha alcuni svantaggi, ma può essere utile quando non sai in anticipo come suddividere il programma in funzioni.
Ti permette di progettare il programma mentre lo sviluppi.

La progettazione di una funzione ha due aspetti:

* L'**interfaccia** descrive come si utilizza la funzione: il nome, i parametri che accetta e ciò che dovrebbe fare.
* L'**implementazione** descrive come la funzione realizza ciò che dovrebbe fare.

Per esempio, ecco la prima versione di `circle` che abbiamo scritto, basata su `polygon`.

````python
def circle(radius):
    circumference = 2 * math.pi * radius
    n = 30
    length = circumference / n
    polygon(n, length)
````

Ed ecco la versione riorganizzata tramite refactoring, che utilizza `arc`.

````python
def circle(radius):
    arc(radius,  360)
````

Queste due funzioni hanno la stessa interfaccia — accettano gli stessi parametri e fanno la stessa cosa — ma hanno implementazioni differenti.

## Docstring

Una **docstring** è una stringa all'inizio di una funzione che ne descrive l'interfaccia (il termine deriva da *documentation*, «documentazione»). Ecco un esempio:

````python
def polyline(n, length, angle):
    """Disegna segmenti della lunghezza indicata, con l’angolo specificato tra loro.

    n: numero intero di segmenti
    length: lunghezza dei segmenti
    angle: angolo tra i segmenti (in gradi)
    """
    for i in range(n):
        forward(length)
        left(angle)
````

Per convenzione, le docstring sono stringhe delimitate da tre virgolette, dette anche **stringhe multilinea**, perché le tre virgolette consentono di estendere una stringa su più righe.

Una docstring dovrebbe:

* spiegare in modo conciso che cosa fa la funzione, senza entrare nei dettagli di come funziona;
* spiegare come ciascun parametro influisce sul comportamento della funzione;
* indicare il tipo previsto per ciascun parametro, se non è evidente.

Scrivere questo tipo di documentazione è una parte importante della progettazione delle interfacce.
Un'interfaccia ben progettata dovrebbe essere semplice da spiegare; se fai fatica a descrivere una delle tue funzioni, forse la sua interfaccia può essere migliorata.

## Debugging

Un'interfaccia è come un contratto tra una funzione e chi la chiama. Chi effettua la chiamata si impegna a fornire determinati argomenti e la funzione si impegna a svolgere determinate operazioni.

Per esempio, `polyline` richiede tre argomenti: `n` deve essere un intero, `length` dovrebbe essere un numero positivo e `angle` deve essere un numero, inteso come misura in gradi.

Questi requisiti si chiamano **precondizioni**, perché devono essere soddisfatti prima che la funzione inizi a essere eseguita. Le condizioni che devono valere alla fine della funzione si chiamano invece **postcondizioni**.
Le postcondizioni comprendono l'effetto previsto della funzione (come disegnare segmenti) ed eventuali effetti collaterali (come spostare la tartaruga o produrre altri cambiamenti).

Le precondizioni sono responsabilità di chi chiama la funzione. Se chi effettua la chiamata viola una precondizione e la funzione non opera correttamente, l'errore è nella chiamata, non nella funzione.

Se le precondizioni sono soddisfatte ma le postcondizioni non lo sono, l'errore si trova nella funzione. Definire chiaramente precondizioni e postcondizioni può aiutare nel debugging.

## Glossario

**progettazione dell'interfaccia (*interface design*):** processo di progettazione dell'interfaccia di una funzione, compresi i parametri che dovrebbe accettare.

**superficie di disegno (*canvas*):** finestra utilizzata per mostrare elementi grafici, come linee, cerchi, rettangoli e altre forme.

**incapsulamento (*encapsulation*):** processo di trasformazione di una sequenza di istruzioni nella definizione di una funzione.

**generalizzazione (*generalization*):** processo di sostituzione di un elemento inutilmente specifico (come un numero) con uno opportunamente generale (come una variabile o un parametro).

**argomento con parola chiave (*keyword argument*):** argomento che include il nome del parametro.

**refactoring:** processo di modifica di un programma funzionante per migliorare le interfacce delle funzioni e altre qualità del codice.

**piano di sviluppo (*development plan*):** procedimento per scrivere programmi.

**docstring:** stringa collocata all'inizio della definizione di una funzione per documentarne l'interfaccia.

**stringa multilinea (*multiline string*):** stringa delimitata da tre virgolette che può estendersi su più righe di programma.

**precondizione (*precondition*):** requisito che chi chiama una funzione deve soddisfare prima dell'esecuzione.

**postcondizione (*postcondition*):** requisito che la funzione deve soddisfare prima di terminare.
