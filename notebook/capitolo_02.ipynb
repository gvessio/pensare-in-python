## Variabili

Una **variabile** è un nome che fa riferimento a un valore. Per creare una variabile possiamo scrivere un'**istruzione di assegnazione** come questa.

````python
n = 17
````

Un'istruzione di assegnazione è composta da tre parti: il nome della variabile a sinistra, il segno di uguale `=` e un'espressione a destra. In questo esempio l'espressione è un numero intero. Nell'esempio seguente è un numero in virgola mobile.

````python
pi = 3.141592653589793
````

Nell'esempio seguente, invece, l'espressione è una stringa.

````python
message = 'And now for something completely different'
````

Quando esegui un'istruzione di assegnazione, non viene mostrato alcun risultato. Python crea la variabile e le assegna un valore, ma l'assegnazione non produce un effetto visibile. Dopo aver creato una variabile, però, puoi utilizzarla come espressione. Possiamo quindi visualizzare il valore di `message` in questo modo:

````python
message
````

Puoi anche utilizzare una variabile all'interno di un'espressione con operatori aritmetici.

````python
n + 25
````

````python
2 * pi
````

E puoi utilizzare una variabile quando chiami una funzione.

````python
round(pi)
````

````python
len(message)
````

## Nomi delle variabili

I nomi delle variabili possono essere lunghi quanto vuoi. Possono contenere lettere e numeri, ma non possono iniziare con un numero. È consentito utilizzare lettere maiuscole, ma per convenzione si usano soltanto lettere minuscole nei nomi delle variabili.

L'unico carattere di punteggiatura ammesso in un nome di variabile è il trattino basso, `_`, spesso utilizzato nei nomi composti da più parole, come `your_name` o `airspeed_of_unladen_swallow`.

Se assegni a una variabile un nome non valido, ottieni un errore di sintassi. Il nome `million!` non è valido perché contiene un segno di punteggiatura.

````python
%%expect SyntaxError

million! = 1000000
````

`76trombones` non è valido perché inizia con un numero.

````python
%%expect SyntaxError

76trombones = 'big parade'
````

Anche `class` non è valido, ma il motivo potrebbe non essere evidente.

````python
%%expect SyntaxError

class = 'Self-Defence Against Fresh Fruit'
````

## L'istruzione `import`

Per utilizzare alcune funzionalità di Python devi **importarle**. Per esempio, l'istruzione seguente importa il modulo `math`.

````python
import math
````

Un **modulo** è una raccolta di variabili e funzioni. Il modulo `math` fornisce una variabile chiamata `pi`, che contiene il valore della costante matematica indicata con $\pi$. Possiamo visualizzarne il valore in questo modo.

````python
math.pi
````

Per utilizzare una variabile contenuta in un modulo, devi inserire l'**operatore punto** (`.`) tra il nome del modulo e quello della variabile.

Il modulo `math` contiene anche funzioni. Per esempio, `sqrt` calcola le radici quadrate.

````python
math.sqrt(25)
````

La funzione `pow`, invece, eleva un numero alla potenza indicata da un secondo numero.

````python
math.pow(5, 2)
````

A questo punto abbiamo visto due modi per elevare un numero a potenza: possiamo usare la funzione `math.pow` oppure l'operatore di elevamento a potenza `**`. Entrambi vanno bene, ma l'operatore è utilizzato più spesso della funzione.

## Espressioni e istruzioni

Finora abbiamo incontrato diversi tipi di espressioni. Un'espressione può essere un singolo valore, come un intero, un numero in virgola mobile o una stringa. Può anche essere una combinazione di valori e operatori, e può includere nomi di variabili e chiamate di funzione. Ecco un'espressione che comprende diversi di questi elementi.

````python
19 + n + round(math.pi) * 2
````

Abbiamo anche incontrato diversi tipi di istruzioni. Un'**istruzione** (*statement*) è un'unità di codice che produce un effetto, ma non ha un valore. Per esempio, un'istruzione di assegnazione crea una variabile e le assegna un valore, ma l'istruzione in sé non ha un valore.

````python
n = 17
````

Allo stesso modo, un'istruzione di importazione produce un effetto — importa un modulo affinché possiamo utilizzare le variabili e le funzioni che contiene — ma non ha un effetto visibile.

````python
import math
````

Calcolare il valore di un'espressione si chiama **valutazione** (*evaluation*). Eseguire un'istruzione si chiama **esecuzione** (*execution*).

## La funzione `print`

Quando valuti un'espressione, il risultato viene visualizzato.

````python
n + 1
````

Ma se valuti più di un'espressione, viene visualizzato soltanto il valore dell'ultima.

````python
n + 2
n + 3
````

Per visualizzare più di un valore puoi utilizzare la funzione `print`.

````python
print(n+2)
print(n+3)
````

Funziona anche con numeri in virgola mobile e stringhe.

````python
print('The value of pi is approximately')
print(math.pi)
````

Puoi anche passare una sequenza di espressioni separate da virgole.

````python
print('The value of pi is approximately', math.pi)
````

Nota che la funzione `print` inserisce uno spazio tra i valori.

## Argomenti

Quando chiami una funzione, l'espressione tra parentesi si chiama **argomento**. Normalmente spiegherei perché si usa questo termine, ma in questo caso il significato tecnico non ha quasi nulla a che vedere con il significato comune della parola, quindi non ci proverò nemmeno.

Alcune delle funzioni viste finora accettano un solo argomento, come `int`.

````python
int('101')
````

Altre ne accettano due, come `math.pow`.

````python
math.pow(5, 2)
````

Alcune funzioni possono accettare ulteriori argomenti facoltativi. Per esempio, `int` può ricevere un secondo argomento che specifica la base del numero.

````python
int('101', 2)
````

La sequenza di cifre `101` in base 2 rappresenta il numero 5 in base 10.

Anche `round` accetta un secondo argomento facoltativo, che indica il numero di cifre decimali a cui arrotondare.

````python
round(math.pi, 3)
````

Alcune funzioni possono accettare un numero qualsiasi di argomenti, come `print`.

````python
print('Any', 'number', 'of', 'arguments')
````

Se chiami una funzione fornendo troppi argomenti, ottieni un `TypeError`.

````python
%%expect TypeError

float('123.0', 2)
````

Se fornisci troppo pochi argomenti, ottieni ancora un `TypeError`.

````python
%%expect TypeError

math.pow(2)
````

E se fornisci un argomento di un tipo che la funzione non può gestire, ottieni di nuovo un `TypeError`.

````python
%%expect TypeError

math.sqrt('123')
````

Questo tipo di controllo può risultare fastidioso quando muovi i primi passi, ma aiuta a individuare e correggere gli errori.

## Commenti

Man mano che i programmi diventano più grandi e complessi, diventano anche più difficili da leggere. I linguaggi formali sono densi e spesso è difficile osservare un frammento di codice e capire che cosa faccia e perché.

Per questo motivo è buona norma aggiungere ai programmi delle note che spieghino, in linguaggio naturale, che cosa sta facendo il programma. Queste note si chiamano **commenti** e iniziano con il simbolo `#`.

````python
# number of seconds in 42:42
seconds = 42 * 60 + 42
````

In questo caso il commento occupa una riga a sé. Puoi anche inserire commenti alla fine di una riga:

````python
miles = 10 / 1.61     # 10 kilometers in miles
````

Tutto ciò che si trova tra `#` e la fine della riga viene ignorato: non ha alcun effetto sull'esecuzione del programma.

I commenti sono particolarmente utili quando documentano aspetti del codice che non sono immediatamente evidenti. È ragionevole presumere che chi legge possa capire *che cosa* fa il codice; è più utile spiegare *perché* lo fa.

Questo commento ripete inutilmente ciò che dice già il codice:

````python
v = 8     # assign 8 to v
````

Questo commento contiene invece informazioni utili che non si trovano nel codice:

````python
v = 8     # velocity in miles per hour
````

Scegliere buoni nomi per le variabili può ridurre la necessità di commenti, ma nomi troppo lunghi possono rendere difficili da leggere le espressioni complesse: occorre trovare un compromesso.

## Debugging

In un programma possono verificarsi tre tipi di errori: errori di sintassi, errori durante l'esecuzione ed errori semantici. Distinguerli è utile per individuarli più rapidamente.

* **Errore di sintassi:** la «sintassi» riguarda la struttura di un programma e le regole che la governano. Se il programma contiene un errore di sintassi, Python non lo esegue e mostra immediatamente un messaggio di errore.

* **Errore durante l'esecuzione (*runtime error*):** se non ci sono errori di sintassi, il programma può iniziare a essere eseguito. Ma se qualcosa va storto, Python mostra un messaggio di errore e si ferma. Questo tipo di errore è chiamato anche **eccezione**, perché segnala che è accaduto qualcosa di eccezionale.

* **Errore semantico:** «semantico» significa relativo al significato. Se il programma contiene un errore semantico, viene eseguito senza mostrare messaggi di errore, ma non fa ciò che volevi. Individuare questi errori può essere difficile, perché bisogna ragionare a ritroso: osservare il risultato e cercare di capire che cosa stia facendo il programma.

Come abbiamo visto, un nome di variabile non valido provoca un errore di sintassi.

````python
%%expect SyntaxError

million! = 1000000
````

Se utilizzi un operatore con un tipo che non supporta, ottieni un errore durante l'esecuzione.

````python
%%expect TypeError

'126' / 3
````

Infine, ecco un esempio di errore semantico. Supponiamo di voler calcolare la media di `1` e `3`, ma di dimenticare l'ordine delle operazioni e scrivere:

````python
1 + 3 / 2
````

Quando questa espressione viene valutata, non produce messaggi di errore: quindi non ci sono errori di sintassi né errori durante l'esecuzione. Tuttavia, il risultato non è la media di `1` e `3`, quindi il programma non è corretto. È un errore semantico: il programma funziona, ma non fa ciò che era previsto.

## Glossario

**variabile:** nome che fa riferimento a un valore.

**istruzione di assegnazione:** istruzione che assegna un valore a una variabile.

**diagramma di stato:** rappresentazione grafica di un insieme di variabili e dei valori a cui fanno riferimento.

**parola chiave:** parola speciale utilizzata per specificare la struttura di un programma.

**istruzione di importazione:** istruzione che legge un file di modulo affinché possiamo utilizzare le variabili e le funzioni che contiene.

**modulo:** file che contiene codice Python, comprese definizioni di funzioni e, talvolta, altre istruzioni.

**operatore punto:** operatore `.` utilizzato per accedere a una funzione di un altro modulo, scrivendo il nome del modulo seguito da un punto e dal nome della funzione.

**valutare:** eseguire le operazioni contenute in un'espressione per calcolarne il valore.

**istruzione:** una o più righe di codice che rappresentano un comando o un'azione.

**eseguire:** eseguire un'istruzione e compiere l'azione che descrive.

**argomento:** valore fornito a una funzione quando viene chiamata.

**commento:** testo inserito in un programma che fornisce informazioni sul programma, ma non influisce sulla sua esecuzione.

**errore durante l'esecuzione:** errore che provoca la visualizzazione di un messaggio e l'interruzione del programma.

**eccezione:** errore rilevato mentre il programma è in esecuzione.

**errore semantico:** errore che induce il programma a fare la cosa sbagliata senza mostrare un messaggio di errore.
