## Definire nuove funzioni

Una **definizione di funzione** specifica il nome di una nuova funzione e la sequenza di istruzioni da eseguire quando viene chiamata. Ecco un esempio:

````python
def print_lyrics():
    print("I'm a lumberjack, and I'm okay.")
    print("I sleep all night and I work all day.")
````

`def` è una parola chiave che indica una definizione di funzione. Il nome della funzione è `print_lyrics`. Qualunque nome valido per una variabile è valido anche per una funzione.

Le parentesi vuote dopo il nome indicano che questa funzione non accetta argomenti.

La prima riga della definizione si chiama **intestazione** (*header*); il resto è il **corpo** (*body*). L'intestazione deve terminare con i due punti e il corpo deve essere rientrato. Per convenzione, si utilizzano quattro spazi. Il corpo di questa funzione contiene due istruzioni `print`; in generale può contenere un numero qualsiasi di istruzioni, di qualunque tipo.

Definire una funzione crea un **oggetto funzione**, che possiamo visualizzare così:

````python
print_lyrics
````

Il risultato indica che `print_lyrics` è una funzione che non accetta argomenti.

Ora che abbiamo definito una funzione, possiamo chiamarla come facciamo con le funzioni predefinite.

````python
print_lyrics()
````

Quando viene eseguita, la funzione esegue le istruzioni del corpo, che visualizzano i primi due versi di *The Lumberjack Song*.

## Parametri

Alcune funzioni che abbiamo visto richiedono argomenti: per esempio, quando chiami `abs`, passi un numero come argomento. Altre accettano più argomenti: `math.pow`, per esempio, ne accetta due, la base e l'esponente.

Ecco la definizione di una funzione che accetta un argomento:

````python
def print_twice(string):
    print(string)
    print(string)
````

Il nome della variabile tra parentesi è un **parametro**. Quando la funzione viene chiamata, il valore dell'argomento viene assegnato al parametro. Per esempio, possiamo chiamare `print_twice` così:

````python
print_twice('Dennis Moore, ')
````

Eseguire questa funzione equivale ad assegnare l'argomento al parametro e poi eseguire il corpo della funzione, come in questo esempio:

````python
string = 'Dennis Moore, '
print(string)
print(string)
````

Puoi anche utilizzare una variabile come argomento.

````python
line = 'Dennis Moore, '
print_twice(line)
````

In questo esempio, il valore di `line` viene assegnato al parametro `string`.

## Chiamare le funzioni

Una volta definita una funzione, puoi utilizzarla all'interno di un'altra funzione. Per dimostrarlo, scriveremo alcune funzioni che visualizzano il testo di [*The Spam Song*](https://www.songfacts.com/lyrics/monty-python/the-spam-song):

> Spam, Spam, Spam, Spam,  
> Spam, Spam, Spam, Spam,  
> Spam, Spam,  
> (Lovely Spam, Wonderful Spam!)  
> Spam, Spam,

Cominciamo dalla funzione seguente, che accetta due parametri.

````python
def repeat(word, n):
    print(word * n)
````

Possiamo utilizzare questa funzione per visualizzare il primo verso della canzone:

````python
spam = 'Spam, '
repeat(spam, 4)
````

Per visualizzare i primi due versi possiamo definire una nuova funzione che utilizza `repeat`.

````python
def first_two_lines():
    repeat(spam, 4)
    repeat(spam, 4)
````

E poi chiamarla così:

````python
first_two_lines()
````

Per visualizzare gli ultimi tre versi possiamo definire un'altra funzione, che utilizza anch'essa `repeat`.

````python
def last_three_lines():
    repeat(spam, 2)
    print('(Lovely Spam, Wonderful Spam!)')
    repeat(spam, 2)
````

````python
last_three_lines()
````

Infine, possiamo riunire tutto in una funzione che visualizza l'intera strofa.

````python
def print_verse():
    first_two_lines()
    last_three_lines()
````

````python
print_verse()
````

Quando eseguiamo `print_verse`, questa chiama `first_two_lines`, che chiama `repeat`, che a sua volta chiama `print`. Sono molte funzioni!

Naturalmente avremmo potuto ottenere lo stesso risultato con meno funzioni, ma lo scopo dell'esempio è mostrare come le funzioni possano collaborare.

## Ripetizione

Se vogliamo visualizzare più di una strofa, possiamo utilizzare un'istruzione `for`. Ecco un semplice esempio.

````python
for i in range(2):
    print(i)
````

La prima riga è un'intestazione che termina con i due punti. La seconda è il corpo, che deve essere rientrato.

L'intestazione inizia con la parola chiave `for`, seguita da una nuova variabile chiamata `i` e dalla parola chiave `in`. Utilizza la funzione `range` per creare una sequenza di due valori: `0` e `1`. In Python, di solito, si comincia a contare da `0`.

Quando l'istruzione `for` viene eseguita, assegna a `i` il primo valore di `range` ed esegue la funzione `print` nel corpo, visualizzando `0`.

Raggiunta la fine del corpo, l'esecuzione torna all'intestazione: per questo l'istruzione si chiama **ciclo** (*loop*). Alla seconda iterazione, assegna a `i` il valore successivo di `range` e lo visualizza. Poiché è l'ultimo valore della sequenza, il ciclo termina.

Ecco come utilizzare un ciclo `for` per visualizzare due strofe della canzone.

````python
for i in range(2):
    print("Verse", i)
    print_verse()
    print()
````

Puoi inserire un ciclo `for` all'interno di una funzione. Per esempio, `print_n_verses` accetta un parametro `n`, che deve essere un intero, e visualizza il numero di strofe richiesto.

````python
def print_n_verses(n):
    for i in range(n):
        print_verse()
        print()
````

In questo esempio non utilizziamo `i` nel corpo del ciclo, ma nell'intestazione deve comunque comparire un nome di variabile.

## Variabili e parametri sono locali

Quando crei una variabile all'interno di una funzione, questa è **locale**: esiste soltanto all'interno della funzione. Per esempio, la funzione seguente accetta due argomenti, li concatena e visualizza due volte il risultato.

````python
def cat_twice(part1, part2):
    cat = part1 + part2
    print_twice(cat)
````

Ecco un esempio di utilizzo:

````python
line1 = 'Always look on the '
line2 = 'bright side of life.'
cat_twice(line1, line2)
````

Quando `cat_twice` viene eseguita, crea una variabile locale chiamata `cat`, che cessa di esistere al termine della funzione. Se proviamo a visualizzarla, otteniamo un `NameError`:

````python
%%expect NameError

print(cat)
````

All'esterno della funzione, `cat` non è definita.

Anche i parametri sono locali. Per esempio, all'esterno di `cat_twice` non esistono `part1` e `part2`.

## Diagrammi dello stack

Per tenere traccia di quali variabili siano accessibili e dove, a volte è utile disegnare un **diagramma dello stack** (*stack diagram*). Come i diagrammi di stato, questi diagrammi mostrano il valore di ciascuna variabile, ma indicano anche a quale funzione appartiene.

Ogni funzione è rappresentata da un **riquadro di attivazione** (*frame*): un riquadro con il nome della funzione all'esterno e i suoi parametri e le variabili locali all'interno.

Ecco il diagramma dello stack per l'esempio precedente.

````python
from diagram import make_frame, Stack

d1 = dict(line1=line1, line2=line2)
frame1 = make_frame(d1, name='__main__', dy=-0.3, loc='left')

d2 = dict(part1=line1, part2=line2, cat=line1+line2)
frame2 = make_frame(d2, name='cat_twice', dy=-0.3,
                    offsetx=0.03, loc='left')

d3 = dict(string=line1+line2)
frame3 = make_frame(d3, name='print_twice',
                    offsetx=0.04, offsety=-0.3, loc='left')

d4 = {"?": line1+line2}
frame4 = make_frame(d4, name='print',
                    offsetx=-0.22, offsety=0, loc='left')

stack = Stack([frame1, frame2, frame3, frame4], dy=-0.8)
````

````python
from diagram import diagram, adjust


width, height, x, y = [3.77, 2.9, 1.1, 2.65]
ax = diagram(width, height)
bbox = stack.draw(ax, x, y)
# adjust(x, y, bbox)

import matplotlib.pyplot as plt
plt.savefig('chap03_stack_diagram.png', dpi=300)
````

I riquadri sono disposti in una pila che indica quale funzione abbia chiamato quale altra funzione. Leggendo dal basso, `print` è stata chiamata da `print_twice`, a sua volta chiamata da `cat_twice`, chiamata da `__main__`: un nome speciale per il livello più esterno. Quando crei una variabile fuori da qualsiasi funzione, questa appartiene a `__main__`.

Nel riquadro di `print`, il punto interrogativo indica che non conosciamo il nome del parametro. Se sei curioso, chiedi a un assistente virtuale: «Quali sono i parametri della funzione `print` di Python?».

## Traceback

Quando si verifica un errore durante l'esecuzione di una funzione, Python mostra il nome della funzione in esecuzione, quello della funzione che l'ha chiamata e così via, risalendo lo stack. Per vedere un esempio, definirò una versione di `print_twice` che contiene un errore: prova a visualizzare `cat`, che è una variabile locale di un'altra funzione.

Il messaggio di errore contiene un **traceback**, che mostra la funzione in esecuzione al momento dell'errore, quella che l'ha chiamata e così via. In questo esempio indica che `cat_twice` ha chiamato `print_twice` e che l'errore si è verificato in `print_twice`.

L'ordine delle funzioni nel traceback è lo stesso dei riquadri nel diagramma dello stack. La funzione che era in esecuzione si trova in fondo.

## Perché usare le funzioni?

Forse non è ancora chiaro perché valga la pena suddividere un programma in funzioni. Ci sono diverse ragioni:

- Creare una funzione permette di dare un nome a un gruppo di istruzioni, rendendo il programma più facile da leggere e da correggere.
- Le funzioni possono rendere un programma più breve, eliminando codice ripetuto. Se in seguito apporti una modifica, devi farlo in un solo punto.
- Suddividere un programma lungo in funzioni consente di verificare e correggere le singole parti, per poi assemblarle in un insieme funzionante.
- Le funzioni ben progettate sono spesso utili in più programmi: dopo averne scritta e verificata una, puoi riutilizzarla.

## Debugging

Il debugging può essere frustrante, ma anche stimolante, interessante e talvolta persino divertente. Ed è una delle abilità più importanti che puoi imparare.

Per certi versi, fare debugging è come svolgere un'indagine: hai degli indizi e devi ricostruire gli eventi che hanno prodotto i risultati osservati.

Il debugging assomiglia anche alla scienza sperimentale. Quando hai un'ipotesi su ciò che non funziona, modifichi il programma e riprovi. Se l'ipotesi è corretta, puoi prevedere il risultato della modifica e avvicinarti a un programma funzionante. Se è sbagliata, devi formularne un'altra.

Per alcune persone programmare e fare debugging sono quasi la stessa cosa: programmare significa correggere gradualmente un programma finché non fa ciò che desideri. L'idea è partire da un programma funzionante e introdurre piccole modifiche, verificandole man mano.

Se trascorri molto tempo a correggere errori, spesso significa che stai scrivendo troppo codice prima di iniziare a provarlo. Procedendo per passi più piccoli, potresti scoprire di andare più veloce.

## Glossario

**definizione di funzione:** istruzione che crea una funzione.

**intestazione (*header*):** prima riga della definizione di una funzione.

**corpo (*body*):** sequenza di istruzioni contenute nella definizione di una funzione.

**oggetto funzione:** valore creato da una definizione di funzione; il nome della funzione è una variabile che fa riferimento a questo oggetto.

**parametro:** nome utilizzato all'interno di una funzione per riferirsi al valore passato come argomento.

**ciclo (*loop*):** istruzione che esegue una o più istruzioni, spesso ripetutamente.

**variabile locale:** variabile definita all'interno di una funzione, accessibile soltanto al suo interno.

**diagramma dello stack:** rappresentazione grafica di una pila di funzioni, delle loro variabili e dei valori a cui fanno riferimento.

**riquadro di attivazione (*frame*):** riquadro di un diagramma dello stack che rappresenta una chiamata di funzione e contiene parametri e variabili locali.

**traceback:** elenco delle funzioni in esecuzione, mostrato quando si verifica un'eccezione.
