## Alcune funzioni restituiscono valori

Quando chiami una funzione come `math.sqrt`, il risultato prende il nome di **valore restituito** (*return value*). Se la chiamata si trova alla fine di una cella, Jupyter visualizza immediatamente il valore restituito.

```python
import math

math.sqrt(42 / math.pi)
```

Se assegni il valore restituito a una variabile, non viene visualizzato.

```python
radius = math.sqrt(42 / math.pi)
```

Ma puoi visualizzarlo in seguito.

```python
radius
```

Oppure puoi usare il valore restituito come parte di un'espressione.

```python
radius + math.sqrt(42 / math.pi)
```

Ecco un esempio di funzione che restituisce un valore.

```python
def circle_area(radius):
    area = math.pi * radius**2
    return area
```

`circle_area` riceve `radius` come parametro e calcola l'area del cerchio con quel raggio.

L'ultima riga è un'istruzione `return`, che restituisce il valore di `area`.

Se chiamiamo la funzione così, Jupyter visualizza il valore restituito.

```python
circle_area(radius)
```

Possiamo assegnare il valore restituito a una variabile.

```python
a = circle_area(radius)
```

Oppure usarlo in un'espressione.

```python
circle_area(radius) + 2 * circle_area(radius / 2)
```

In seguito possiamo visualizzare il valore della variabile a cui abbiamo assegnato il risultato.

```python
a
```

Non possiamo però accedere a `area`.

```python
%%expect NameError

area
```

`area` è una variabile locale della funzione: non possiamo accedervi dall'esterno.

## Altre restituiscono `None`

## Valori restituiti e condizioni

Se Python non fornisse `abs`, potremmo scriverla così.

```python
def absolute_value(x):
    if x < 0:
        return -x
    else:
        return x
```

Se `x` è negativo, la prima istruzione `return` restituisce `-x` e termina immediatamente la funzione. Altrimenti, la seconda restituisce `x` e termina la funzione. Questa versione è quindi corretta.

Quando inserisci istruzioni `return` in un blocco condizionale, devi però assicurarti che ogni possibile percorso di esecuzione raggiunga un `return`. Ecco una versione sbagliata di `absolute_value`.

```python
def absolute_value_wrong(x):
    if x < 0:
        return -x
    if x > 0:
        return x
```

Ecco cosa succede se chiamiamo la funzione passando `0`.

```python
absolute_value_wrong(0)
```

Non compare nulla! Quando `x` vale `0`, nessuna delle due condizioni è vera e la funzione termina senza eseguire un `return`: restituisce quindi `None`, che Jupyter non visualizza.

Ecco un altro esempio: una versione di `absolute_value` con un `return` aggiuntivo alla fine.

```python
def absolute_value_extra_return(x):
    if x < 0:
        return -x
    else:
        return x

    return 'This is dead code'
```

Se `x` è negativo, viene eseguito il primo `return` e la funzione termina. Altrimenti viene eseguito il secondo `return` e la funzione termina. In entrambi i casi non raggiungiamo mai il terzo `return`.

Il codice che non può mai essere eseguito si chiama **codice morto** (*dead code*). Di solito non provoca danni, ma spesso rivela un fraintendimento e può confondere chi legge il programma.

## Sviluppo incrementale

Quando scrivi funzioni più grandi, potresti impiegare più tempo nel debugging. Per affrontare programmi sempre più complessi, puoi provare lo **sviluppo incrementale**: aggiungere e verificare piccole quantità di codice per volta.

Supponiamo di voler calcolare la distanza tra due punti di coordinate $(x_1, y_1)$ e $(x_2, y_2)$. Per il teorema di Pitagora:

$$\mathrm{distance} = \sqrt{(x_2 - x_1)^2 + (y_2 - y_1)^2}$$

Il primo passo è stabilire come debba essere la funzione Python `distance`: quali sono gli ingressi (parametri) e quale l'uscita (valore restituito)?

Gli ingressi sono le coordinate dei punti; il valore restituito è la distanza. Possiamo già scrivere uno scheletro della funzione:

```python
def distance(x1, y1, x2, y2):
    return 0.0
```

Questa versione non calcola ancora le distanze: restituisce sempre zero. È però una funzione completa, con un valore restituito, e possiamo provarla prima di complicarla.

Per verificare la nuova funzione, la chiamiamo con argomenti di esempio:

```python
distance(1, 2, 4, 6)
```

Ho scelto questi valori in modo che la distanza orizzontale sia `3` e quella verticale `4`. Il risultato corretto è quindi `5`, l'ipotenusa di un triangolo rettangolo `3-4-5`. Quando si verifica una funzione, è utile conoscere la risposta giusta.

Abbiamo confermato che la funzione viene eseguita e restituisce un valore: possiamo iniziare ad aggiungere codice al corpo. Un buon passo successivo è calcolare `x2 - x1` e `y2 - y1`. Ecco una versione che memorizza questi valori in variabili temporanee e li visualizza.

```python
def distance(x1, y1, x2, y2):
    dx = x2 - x1
    dy = y2 - y1
    print('dx is', dx)
    print('dy is', dy)
    return 0.0
```

Se funziona, la funzione dovrebbe visualizzare `dx is 3` e `dy is 4`. Sappiamo così che riceve gli argomenti corretti ed esegue bene il primo calcolo. In caso contrario, ci sono soltanto poche righe da controllare.

```python
distance(1, 2, 4, 6)
```

Fin qui tutto bene. Calcoliamo ora la somma dei quadrati di `dx` e `dy`:

```python
def distance(x1, y1, x2, y2):
    dx = x2 - x1
    dy = y2 - y1
    dsquared = dx**2 + dy**2
    print('dsquared is: ', dsquared)
    return 0.0
```

Possiamo eseguire nuovamente la funzione e controllare il risultato, che dovrebbe essere `25`.

```python
distance(1, 2, 4, 6)
```

Infine, possiamo usare `math.sqrt` per calcolare la distanza:

```python
def distance(x1, y1, x2, y2):
    dx = x2 - x1
    dy = y2 - y1
    dsquared = dx**2 + dy**2
    result = math.sqrt(dsquared)
    print("result is", result)
```

E verificarla.

```python
distance(1, 2, 4, 6)
```

Il risultato è corretto, ma questa versione visualizza il risultato anziché restituirlo: il valore restituito è quindi `None`.

Possiamo correggerla sostituendo `print` con un'istruzione `return`.

```python
def distance(x1, y1, x2, y2):
    dx = x2 - x1
    dy = y2 - y1
    dsquared = dx**2 + dy**2
    result = math.sqrt(dsquared)
    return result
```

Questa versione di `distance` è una funzione pura. Se la chiamiamo così, viene visualizzato soltanto il risultato.

```python
distance(1, 2, 4, 6)
```

Se invece assegniamo il risultato a una variabile, non viene visualizzato nulla.

```python
d = distance(1, 2, 4, 6)
```

Le istruzioni `print` che abbiamo scritto sono utili per il debugging, ma possiamo rimuoverle quando la funzione funziona. Questo codice di supporto si chiama **impalcatura** (*scaffolding*): aiuta a costruire il programma, ma non fa parte del prodotto finale.

L'esempio illustra lo sviluppo incrementale. Gli aspetti fondamentali sono:

1. Parti da un programma funzionante, apporta piccole modifiche e verifica dopo ogni modifica.
2. Usa variabili per conservare valori intermedi, così da poterli visualizzare e controllare.
3. Quando il programma funziona, rimuovi l'impalcatura.

Se si verifica un errore, dovresti avere una buona idea di dove cercarlo. Lo sviluppo incrementale può farti risparmiare molto tempo di debugging.

## Funzioni booleane

Le funzioni possono restituire i valori booleani `True` e `False`. È spesso utile per racchiudere un controllo complesso in una funzione. Per esempio, `is_divisible` verifica se `x` è divisibile per `y` senza resto.

```python
def is_divisible(x, y):
    if x % y == 0:
        return True
    else:
        return False
```

Ecco come si usa.

```python
is_divisible(6, 4)
```

```python
is_divisible(6, 3)
```

All'interno della funzione, il risultato dell'operatore `==` è già un booleano: possiamo quindi scrivere la funzione in modo più conciso, restituendolo direttamente.

```python
def is_divisible(x, y):
    return x % y == 0
```

Le funzioni booleane vengono spesso utilizzate nelle istruzioni condizionali.

```python
if is_divisible(6, 2):
    print('divisible')
```

Potrebbe venirti voglia di scrivere qualcosa del genere:

```python
if is_divisible(6, 2) == True:
    print('divisible')
```

Ma il confronto è superfluo.

## Ricorsione con valori restituiti

Ora che sappiamo scrivere funzioni che restituiscono valori, possiamo scrivere funzioni ricorsive che restituiscono valori. Abbiamo così superato una soglia importante: il sottoinsieme di Python che conosciamo è ora **Turing-completo**, cioè può eseguire qualsiasi calcolo descrivibile mediante un algoritmo.

Per illustrare la ricorsione con valori restituiti, calcoleremo alcune funzioni matematiche definite ricorsivamente. Una definizione ricorsiva somiglia a una definizione circolare, perché richiama ciò che sta definendo. Una definizione davvero circolare non è molto utile:

> vorpal: aggettivo usato per descrivere qualcosa che è vorpal.

Se trovassi questa definizione nel dizionario, potresti irritarti. Se invece cercassi la definizione di fattoriale, indicato dal simbolo $!$, potresti trovare:

$$\begin{aligned}
0! &= 1 \\
n! &= n~(n-1)!
\end{aligned}$$

Questa definizione dice che il fattoriale di $0$ è $1$ e quello di ogni altro valore $n$ è $n$ moltiplicato per il fattoriale di $n-1$.

Se sai scrivere una definizione ricorsiva, puoi scrivere un programma Python per calcolarla. Seguendo lo sviluppo incrementale, iniziamo da una funzione che riceve `n` e restituisce sempre `0`.

```python
def factorial(n):
    return 0
```

Aggiungiamo la prima parte della definizione: se l'argomento è `0`, basta restituire `1`:

```python
def factorial(n):
    if n == 0:
        return 1
    else:
        return 0
```

Completiamo la seconda parte: se `n` non è `0`, dobbiamo chiamare ricorsivamente la funzione per calcolare il fattoriale di `n-1` e moltiplicare il risultato per `n`:

```python
def factorial(n):
    if n == 0:
        return 1
    else:
        recurse = factorial(n-1)
        return n * recurse
```

Il flusso di esecuzione è simile a quello di `countdown` nel Capitolo 5. Se chiamiamo `factorial` con `3`:

Poiché `3` non è `0`, seguiamo il secondo ramo e calcoliamo il fattoriale di `n-1`…

> Poiché `2` non è `0`, seguiamo il secondo ramo e calcoliamo il fattoriale di `n-1`…
>
> > Poiché `1` non è `0`, seguiamo il secondo ramo e calcoliamo il fattoriale di `n-1`…
> >
> > > Poiché `0` è uguale a `0`, seguiamo il primo ramo e restituiamo `1`, senza altre chiamate ricorsive.
> >
> > Il valore restituito `1` viene moltiplicato per `n`, che vale `1`, e il risultato viene restituito.
>
> Il valore restituito `1` viene moltiplicato per `n`, che vale `2`, e il risultato viene restituito.

Il valore restituito `2` viene moltiplicato per `n`, che vale `3`, e il risultato `6` diventa il valore restituito dalla chiamata che ha avviato tutto il processo.

La figura seguente mostra il diagramma dello stack per questa sequenza di chiamate.

```python
from diagram import Frame, Stack, make_binding

main = Frame([], name='__main__', loc='left')
frames = [main]

ns = 3, 2, 1
recurses = 2, 1, 1
results = 6, 2, 1

for n, recurse, result in zip(ns, recurses, results):
    binding1 = make_binding('n', n)
    binding2 = make_binding('recurse', recurse)
    frame = Frame([binding1, binding2],
                  name='factorial', value=result,
                  loc='left', dx=1.2)
    frames.append(frame)

binding1 = make_binding('n', 0)
frame = Frame([binding1], name='factorial', value=1,
              shim=1.2, loc='left', dx=1.4)
frames.append(frame)

stack = Stack(frames, dy=-0.45)
```

```python
from diagram import diagram, adjust

width, height, x, y = [2.74, 2.26, 0.73, 2.05]
ax = diagram(width, height)
bbox = stack.draw(ax, x, y)
# adjust(x, y, bbox)
```

I valori restituiti risalgono lo stack. In ogni frame, il valore restituito è il prodotto di `n` e `recurse`.

Nell'ultimo frame, la variabile locale `recurse` non esiste perché il ramo che la crea non viene eseguito.

## Un atto di fiducia

Seguire il flusso di esecuzione è un modo per leggere i programmi, ma può diventare rapidamente complicato. Un'alternativa è quella che chiamo «atto di fiducia» (*leap of faith*). Quando incontri una chiamata di funzione, invece di seguirne l'esecuzione, **supponi** che la funzione funzioni correttamente e restituisca il risultato giusto.

In realtà lo fai già con le funzioni predefinite: quando chiami `abs` o `math.sqrt`, non ne esamini il codice, ma presumi che funzionino.

Lo stesso vale per le funzioni che scrivi tu. Per esempio, abbiamo scritto `is_divisible` per stabilire se un numero è divisibile per un altro. Una volta convinti che sia corretta, possiamo usarla senza rileggerne il corpo.

Lo stesso principio vale per la ricorsione. Quando arrivi alla chiamata ricorsiva, supponi che funzioni e chiediti: «Se so calcolare il fattoriale di $n-1$, posso calcolare quello di $n$?». La definizione ricorsiva dice di sì: basta moltiplicare per $n$.

Naturalmente è un po' strano supporre che una funzione funzioni prima ancora di aver finito di scriverla, ma è proprio per questo che si parla di un atto di fiducia!

## Fibonacci

Dopo `factorial`, l'esempio più comune di funzione ricorsiva è `fibonacci`, definita così:

$$\begin{aligned}
\mathrm{fibonacci}(0) &= 0 \\
\mathrm{fibonacci}(1) &= 1 \\
\mathrm{fibonacci}(n) &= \mathrm{fibonacci}(n-1) + \mathrm{fibonacci}(n-2)
\end{aligned}$$

Tradotta in Python, diventa:

```python
def fibonacci(n):
    if n == 0:
        return 0
    elif  n == 1:
        return 1
    else:
        return fibonacci(n-1) + fibonacci(n-2)
```

Se provi a seguire il flusso di esecuzione, anche per valori piccoli di $n$, ti esplode la testa. Ma, secondo il principio dell'atto di fiducia, se supponi che le due chiamate ricorsive funzionino correttamente, puoi essere certo che l'ultima istruzione `return` sia corretta.

Per inciso, questo modo di calcolare i numeri di Fibonacci è molto inefficiente. Nel [Capitolo 10](https://allendowney.github.io/ThinkPython/chap10.html) spiegherò perché e suggerirò come migliorarlo.

## Controllo dei tipi

Che cosa succede se chiamiamo `factorial` passando `1.5` come argomento?

```python
%%expect RecursionError

factorial(1.5)
```

Sembra una ricorsione infinita. Com'è possibile? La funzione ha casi base quando `n == 1` o `n == 0`. Ma se `n` non è intero, possiamo *saltare* il caso base e continuare a chiamare la funzione.

In questo esempio `n` parte da `1.5`. Alla prima chiamata ricorsiva diventa `0.5`, poi `-0.5`. Da lì continua a diminuire, ma non sarà mai `0`.

Per evitare la ricorsione infinita, possiamo usare la funzione predefinita `isinstance` per controllare il tipo dell'argomento. Ecco come verificare se un valore è un intero.

```python
isinstance(3, int)
```

```python
isinstance(1.5, int)
```

Ecco una versione di `factorial` con controllo degli errori.

```python
def factorial(n):
    if not isinstance(n, int):
        print('factorial is only defined for integers.')
        return None
    elif n < 0:
        print('factorial is not defined for negative numbers.')
        return None
    elif n == 0:
        return 1
    else:
        return n * factorial(n-1)
```

Prima verifica che `n` sia un intero. In caso contrario, mostra un messaggio di errore e restituisce `None`.

```python
factorial('crunchy frog')
```

Poi controlla se `n` è negativo. In tal caso, mostra un messaggio di errore e restituisce `None`.

```python
factorial(-2)
```

Se superiamo entrambi i controlli, sappiamo che `n` è un intero non negativo e possiamo essere sicuri che la ricorsione terminerà. Controllare i parametri di una funzione per verificare che abbiano tipi e valori corretti si chiama **validazione degli input**.

## Debugging

Suddividere un programma grande in funzioni più piccole crea punti di controllo naturali per il debugging. Se una funzione non funziona, ci sono tre possibilità:

- C'è un problema con gli argomenti ricevuti: una **precondizione** è violata.
- C'è un problema nella funzione: una **postcondizione** è violata.
- Il codice che chiama la funzione usa male il valore restituito.

Per escludere la prima possibilità, puoi aggiungere un `print` all'inizio della funzione per mostrare i parametri (ed eventualmente i loro tipi), oppure verificare esplicitamente le precondizioni.

Se i parametri sembrano corretti, puoi aggiungere un `print` prima di ogni `return` per mostrare il valore restituito. Se possibile, chiama la funzione con argomenti che rendano semplice controllare il risultato.

Se la funzione sembra funzionare, controlla la chiamata per assicurarti che il valore restituito sia usato correttamente, o almeno che venga usato!

Aggiungere `print` all'inizio e alla fine di una funzione può rendere più visibile il flusso di esecuzione. Ecco una versione di `factorial` con istruzioni di stampa:

```python
def factorial(n):
    space = ' ' * (4 * n)
    print(space, 'factorial', n)
    if n == 0:
        print(space, 'returning 1')
        return 1
    else:
        recurse = factorial(n-1)
        result = n * recurse
        print(space, 'returning', result)
        return result
```

`space` è una stringa di spazi che controlla il rientro del testo visualizzato. Ecco il risultato di `factorial(3)`:

```python
factorial(3)
```

Se il flusso di esecuzione ti confonde, questo tipo di output può aiutarti. Serve un po' di pratica per costruire un'impalcatura efficace, ma anche una piccola impalcatura può risparmiare molto debugging.

## Glossario

**valore restituito (*return value*):** risultato di una funzione. Se la chiamata è usata come espressione, il valore restituito è il valore dell'espressione.

**funzione pura (*pure function*):** funzione che non visualizza nulla e non produce altri effetti, oltre a restituire un valore.

**codice morto (*dead code*):** parte di un programma che non può mai essere eseguita, spesso perché si trova dopo un'istruzione `return`.

**sviluppo incrementale (*incremental development*):** metodo di sviluppo che cerca di evitare il debugging aggiungendo e verificando piccole quantità di codice per volta.

**impalcatura (*scaffolding*):** codice usato durante lo sviluppo, ma assente dalla versione finale.

**Turing-completo (*Turing complete*):** un linguaggio, o un suo sottoinsieme, è Turing-completo se può eseguire qualsiasi calcolo descrivibile da un algoritmo.

**validazione degli input (*input validation*):** controllo dei parametri di una funzione per verificare che abbiano tipi e valori corretti.
