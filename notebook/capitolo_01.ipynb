## Operatori aritmetici

Un **operatore aritmetico** è un simbolo che rappresenta un calcolo. Per esempio, il segno `+` esegue l'addizione.

```python
30 + 12
```

Il segno meno `-` è l’operatore della sottrazione.

```python
43 - 1
```

L’asterisco `*` esegue la moltiplicazione.

```python
6 * 7
```

La barra `/` esegue la divisione.

```python
84 / 2
```

Osserva che il risultato della divisione è `42.0` e non `42`. Questo accade perché Python distingue due tipi di numeri:

- Gli **interi** (*integers*) rappresentano numeri senza parte frazionaria o decimale.
- I **numeri in virgola mobile** (*floating-point numbers*) rappresentano numeri interi e numeri con una parte decimale.

Se sommi, sottrai o moltiplichi due interi, il risultato è un intero. Se dividi due interi con `/`, invece, il risultato è un numero in virgola mobile.

Python mette a disposizione anche `//`, che esegue la **divisione intera**. Il risultato della divisione intera di due interi è sempre un intero.

```python
84 // 2
```

La divisione intera è detta anche *floor division*, perché arrotonda sempre verso il basso, cioè verso il valore intero inferiore. Prova con un risultato non esatto.

```python
85 // 2
```

Infine, l’operatore `**` esegue l’elevamento a potenza.

```python
7 ** 2
```

In alcuni altri linguaggi il simbolo `^` rappresenta l'elevamento a potenza. In Python, però, è un operatore sui bit chiamato **XOR**. Se non conosci gli operatori bit a bit, il risultato potrebbe sorprenderti.

```python
7 ^ 2
```

Non approfondiremo gli operatori sui bit in questo percorso; puoi consultarne una [breve panoramica](https://wiki.python.org/moin/BitwiseOperators).

## Espressioni

Una combinazione di numeri e operatori si chiama **espressione**. Un'espressione può contenere un numero qualsiasi di operatori e numeri. Ecco un'espressione con due operatori.

```python
6 + 6 ** 2
```

Nota che l'elevamento a potenza viene eseguito prima dell'addizione. Python segue l'ordine delle operazioni che potresti aver studiato a scuola: prima le potenze, poi moltiplicazioni e divisioni, infine addizioni e sottrazioni.

Nel prossimo esempio la moltiplicazione precede l'addizione.

```python
12 + 5 * 6
```

Se vuoi che l’addizione venga eseguita per prima, puoi usare le parentesi.

```python
(12 + 5) * 6
```

Ogni espressione ha un **valore**. Per esempio, il valore dell’espressione `6 * 7` è `42`.

## Funzioni aritmetiche

Oltre agli operatori, Python mette a disposizione alcune **funzioni** che lavorano con i numeri. Per esempio, `round` riceve un numero in virgola mobile e lo arrotonda all'intero più vicino.

```python
round(42.4)
```

```python
round(42.6)
```

La funzione `abs` calcola il valore assoluto di un numero. Per un numero positivo, il valore assoluto coincide con il numero stesso.

```python
abs(42)
```

Per un numero negativo, il valore assoluto è positivo.

```python
abs(-42)
```

Quando usiamo una funzione in questo modo, diciamo che la stiamo **chiamando**. Un'espressione che chiama una funzione è una **chiamata di funzione**.

Le parentesi sono obbligatorie. Se le ometti, ottieni un messaggio di errore. La prossima cella produce **intenzionalmente** un errore di sintassi: non significa che il notebook non funzioni.

```python
%%expect SyntaxError
abs 42
```

Puoi ignorare la prima riga del messaggio, che per ora non ci serve. La riga successiva mostra il codice errato e un accento circonflesso (`^`) indica il punto in cui è stato individuato il problema.

L'ultima riga segnala un **errore di sintassi** (*SyntaxError*): la struttura dell'espressione non è corretta. In questo caso mancano le parentesi richieste dalla chiamata di funzione.

Vediamo che cosa succede omettendo sia le parentesi sia il valore.

```python
abs
```

Il nome di una funzione, da solo, è un’espressione valida e ha un valore. Quando Python lo visualizza, indica che `abs` è una funzione e mostra altre informazioni che chiariremo in seguito.

## Stringhe

Oltre ai numeri, Python può rappresentare sequenze di lettere, chiamate **stringhe** (*strings*), come se i caratteri fossero infilati uno dopo l'altro su un filo. Per scrivere una stringa possiamo racchiudere una sequenza di caratteri tra apici semplici diritti.

```python
'Hello'
```

Sono validi anche i doppi apici.

```python
"world"
```

I doppi apici permettono di scrivere facilmente una stringa contenente un apostrofo, che usa lo stesso simbolo degli apici semplici.

```python
"it's a small "
```

Le stringhe possono contenere anche spazi, punteggiatura e cifre.

```python
'Well, '
```

L'operatore `+` funziona anche con le stringhe: ne unisce due in una sola. Questa operazione si chiama **concatenazione**.

```python
'Well, ' + "it's a small " + 'world.'
```

Anche `*` funziona con le stringhe: ne crea più copie e le concatena.

```python
'Spam, ' * 4
```

Gli altri operatori aritmetici non funzionano con le stringhe. Python offre invece la funzione `len`, che calcola la lunghezza di una stringa.

```python
len('Spam')
```

Nota che `len` conta le lettere tra gli apici, ma non gli apici stessi. Quando crei una stringa, assicurati di usare virgolette diritte. Il simbolo chiamato *backtick* provoca un errore di sintassi.

```python
%%expect SyntaxError
`Hello`
```

Anche le virgolette tipografiche curve non sono valide come delimitatori di stringhe.

```python
%%expect SyntaxError
‘Hello’
```

## Valori e tipi

Finora abbiamo incontrato tre categorie di valori:

- `2` è un numero intero;
- `42.0` è un numero in virgola mobile;
- `'Hello'` è una stringa.

Una categoria di valori si chiama **tipo** (*type*). Ogni valore ha un tipo, o appartiene a un tipo.

La funzione `type` permette di scoprire il tipo di qualsiasi valore. Il tipo di un intero è `int`.

```python
type(2)
```

Il tipo di un numero in virgola mobile è `float`.

```python
type(42.0)
```

Il tipo di una stringa è `str`.

```python
type('Hello, World!')
```

I tipi `int`, `float` e `str` possono essere usati come funzioni. Per esempio, `int` può ricevere un numero in virgola mobile e convertirlo in un intero, troncando la parte decimale (ossia arrotondando verso zero).

```python
int(42.9)
```

E `float` può convertire un intero in un numero in virgola mobile.

```python
float(42)
```

Ecco qualcosa che può creare confusione: che cosa succede se racchiudi una sequenza di cifre tra apici?

```python
'126'
```

Sembra un numero, ma è una stringa.

```python
type('126')
```

Se provi a usarla come un numero, potresti ottenere un errore.

```python
%%expect TypeError
'126' / 3
```

Questo esempio genera un `TypeError`: i valori su cui agisce l'operatore, detti **operandi**, hanno tipi incompatibili con l'operazione. Il messaggio indica che `/` non supporta in questo caso gli operandi di tipo `str` e `int`.

Se una stringa contiene cifre, puoi usare `int` per convertirla in un intero.

```python
int('126') / 3
```

Se una stringa contiene cifre e un punto decimale, puoi usare `float` per convertirla in un numero in virgola mobile.

```python
float('12.6')
```

Quando scrivi un intero grande, potresti essere tentato di separare le migliaia con virgole, per esempio `1,000,000`. In Python è un'espressione valida, ma non produce un singolo intero.

```python
1,000,000
```

Python interpreta questa espressione come una sequenza di interi separati da virgole. Studieremo più avanti questo tipo di sequenza. Per rendere più leggibili i numeri grandi, puoi usare il trattino basso `_`.

```python
1_000_000
```

## Linguaggi formali e naturali

I **linguaggi naturali** sono le lingue parlate dalle persone, come italiano, inglese, spagnolo e francese. Non sono stati progettati deliberatamente: si sono evoluti nel tempo.

I **linguaggi formali**, invece, sono progettati per applicazioni specifiche. Per esempio, la notazione matematica è un linguaggio formale particolarmente adatto a esprimere relazioni tra numeri e simboli. Analogamente, i linguaggi di programmazione sono progettati per esprimere calcoli.

Pur condividendo alcune caratteristiche, i linguaggi naturali e quelli formali differiscono in modi importanti:

- **Ambiguità.** I linguaggi naturali sono pieni di ambiguità, che normalmente risolviamo grazie al contesto e ad altre informazioni. I linguaggi formali sono progettati per essere quasi o completamente privi di ambiguità: un programma deve avere un significato determinato indipendentemente dal contesto.
- **Ridondanza.** Per compensare l'ambiguità e ridurre le incomprensioni, le lingue naturali usano spesso espressioni ridondanti e prolisse. I linguaggi formali sono meno ridondanti e più concisi.
- **Letteralità.** Le lingue naturali sono ricche di modi di dire e metafore. I linguaggi formali significano esattamente ciò che esprimono.

Poiché cresciamo parlando lingue naturali, adattarci ai linguaggi formali può essere difficile. Sono più densi e richiedono più tempo per essere letti. Inoltre, la loro struttura è importante: non sempre conviene leggerli semplicemente dall'alto in basso e da sinistra a destra. Infine, i dettagli contano. Piccoli errori di ortografia o punteggiatura, tollerabili in una lingua naturale, possono cambiare radicalmente il significato di un programma.

## Debugging

I programmatori commettono errori. Per ragioni scherzose, gli errori di programmazione sono chiamati **bug**, mentre il processo di individuarli e correggerli si chiama **debugging**.

La programmazione, e soprattutto il debugging, può suscitare emozioni intense. Se stai affrontando un bug difficile, potresti sentirti arrabbiato, triste o in imbarazzo.

Prepararsi a queste reazioni può aiutare a gestirle. Un possibile approccio è pensare al computer come a un dipendente con determinati punti di forza, come la velocità e la precisione, e particolari debolezze, come la mancanza di empatia e l'incapacità di cogliere il quadro generale.

Il tuo compito è essere un buon responsabile: trovare modi per sfruttare i punti di forza e mitigare le debolezze. E trovare modi per usare le emozioni per impegnarti nel problema, senza lasciare che le tue reazioni interferiscano con la capacità di lavorare efficacemente.

## Glossario

- **Operatore aritmetico:** simbolo, come `+` o `*`, che indica un'operazione aritmetica.
- **Intero:** tipo numerico che rappresenta valori senza parte frazionaria.
- **Numero in virgola mobile:** tipo numerico che rappresenta numeri interi e numeri con parte decimale.
- **Divisione intera:** operatore `//`, che divide due numeri e arrotonda il risultato verso il basso a un intero.
- **Espressione:** combinazione di variabili, valori e operatori.
- **Valore:** un intero, un numero in virgola mobile, una stringa o un altro oggetto che incontreremo.
- **Funzione:** sequenza di istruzioni con un nome che svolge un'operazione utile; può ricevere argomenti e produrre un risultato, oppure no.
- **Chiamata di funzione:** espressione che esegue una funzione; comprende il nome seguito da una lista di argomenti tra parentesi.
- **Errore di sintassi:** errore che impedisce di analizzare correttamente un programma e quindi di eseguirlo.
- **Stringa:** tipo che rappresenta sequenze di caratteri.
- **Concatenazione:** unione di due stringhe, una dopo l'altra.
- **Tipo:** categoria di valori; finora `int`, `float` e `str`.
- **Operando:** valore sul quale agisce un operatore.
- **Linguaggio naturale:** lingua parlata dalle persone, evolutasi naturalmente.
- **Linguaggio formale:** linguaggio progettato per scopi specifici, come rappresentare idee matematiche o programmi informatici. Tutti i linguaggi di programmazione sono linguaggi formali.
- **Bug:** errore in un programma.
- **Debugging:** processo di ricerca e correzione degli errori.
